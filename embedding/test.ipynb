```python
import numpy as np
```

```python
text1 = "من سیب دوست دارم"
text2 = "او سیب دوست ندارد"

# ۲. نرمال‌سازی ساده (شکستن متن به کلمات)
words1 = text1.split()  # ['من', 'سیب', 'دوست', 'دارم']
words2 = text2.split()  # ['او', 'سیب', 'دوست', 'ندارد']

# ۳. ساخت لغت‌نامه (Vocabulary): جمع‌آوری تمام کلمات منحصربه‌فرد
# از مجموعه (set) استفاده می‌کنیم تا کلمات تکراری حذف شوند
vocabulary = list(set(words1 + words2))
print("لغت‌نامه مرجع:", vocabulary)
# خروجی لغت‌نامه چیزی شبیه به این خواهد بود: ['او', 'ندارد', 'سیب', 'دارم', 'من', 'دوست']

# ۴. تابع تبدیل متن به بردار (بر اساس لغت‌نامه)
def text_to_vector(text_words, vocab):
    vector = []
    for word in vocab:
        # اگر کلمه در متن بود، تعداد تکرار آن را بگذار، وگرنه صفر بگذار
        count = text_words.count(word)
        vector.append(count)
    return vector

# ۵. تبدیل متن‌ها به بردار عددی
vector1 = text_to_vector(words1, vocabulary)
vector2 = text_to_vector(words2, vocabulary)

print("بردار متن اول:", vector1)
print("بردار متن دوم:", vector2)

# ۶. محاسبه ضرب داخلی (Dot Product) به روش دستی و ریاضی
# ضرب مولفه به مولفه و جمع آن‌ها با هم
dot_product = sum(x * y for x, y in zip(vector1, vector2))
print("حاصل ضرب داخلی دو بردار:", dot_product)
```

```python
# tf and idf termonology this is much better then previous alogorithm that was bag of words

import math 

corpus =[
    "he likes apples",
    "he does not like apples",
    "it is raining the weather is cold"
]

print(len(corpus))
docs_word= [doc.split() for doc in corpus]
print(docs_word)
vacablary= list(set(word for doc in docs_word for word in doc))
print(vacablary)


total_documents= len(corpus)
idf_dic= {}

for word in vacablary:
        docs_with_words= sum(1 for doc in docs_word if word in doc )
        idf_dic[word]=  math.log(total_documents / docs_with_words) 
def get_tfidf_vector(text_word, vocab, idf_table):
    vector= []
    total_word_in_doc= len(text_word)  
    
    for word in vocab:
        if word in vocab:
            tf= text_word.count(word) / total_word_in_doc
            tfidf_weight= tf * idf_table[word]
            vector.append(tfidf_weight)
        else:
            vector.append(0.0)      

    return vector 

vector1 = get_tfidf_vector(docs_word[0], vacablary, idf_dic)
vector2 = get_tfidf_vector(docs_word[1], vacablary, idf_dic)

```

```python

print("vacablary:", vacablary)
print("the first doc of tf_idf", [round(num, 4) for num in vector1])
print("the second doc of tf_idf", [round(num, 4) for num in vector2])

dot_product = sum(x * y for x, y in zip(vector1, vector2))
print("the dot product of TF_IDF:", round(dot_product, 4))
```
