```python
def find_missing_number(array):
    for i in range(0,len(array)-1,1):
        if array[i] + 1 != (array[i+1]):
            print(array[i] + 1)
        
print(find_missing_number([1,3,4,6]))        
        
```

```python
def second_largest_number(array):
    first_max = float('-inf')
    second_max = float('-inf')
    for i in array:
        if (first_max < i):
            second_max=first_max
            first_max=i
        elif(first_max > i and second_max < i):
            second_max=i   
    return second_max
print(second_largest_number([4,5,3,2,6,3]))       
```

```python
def isPalendrome(string):
    reverse= ""
    for i in range(len(string),0,-1):
        reverse += string[i-1]
    return (reverse == string)    
print(isPalendrome("EHE"))        
        
```

```python
def is_palendrome(text):
    text= text.replace(" ", "").lower()
    return text == text[::-1]

print(is_palendrome("Ehsanullah"))
```

```python
def is_palindrome(text):
    text = text.replace(" ", "").lower()
    reverse = text[::-1]
    return text == reverse
```
