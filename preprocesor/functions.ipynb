```python
def mod2_plus_5(x1,x2,x3):
    
    newX1= x1 % 2 + 5
    newX2= x2 % 2 + 5
    newX4= x3 % 2 + 5
    
    return (newX1,newX2, newX4)

mod2_plus_5(3,4,5)
```

```python
def mod2_plus_5_inner_function(x1,x2,x3):
    
    def inner_function(x):
        return x % 2 + 5
    
    return (inner_function(x1),inner_function(x2),inner_function(x3))

mod2_plus_5_inner_function(32,4,2)
```

```python
def start_cart():
    total_price = 0
    
    def add_item(item_name, price):
        nonlocal total_price
        total_price += price
        return f"you add {item_name} and your total price become: {total_price} "
        
    return add_item

my_cart = start_cart()

print(my_cart("book", 500))  
print(my_cart("pen", 50))   
```
