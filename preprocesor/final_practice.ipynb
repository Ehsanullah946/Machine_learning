```python
df2 = pd.read_csv("tested.csv")
df2.tail()
```

```python
df2["Age"]= df2["Age"].fillna(df2["Age"].mean())
df2.tail()
```

```python
# maping sex with 1 and 0 
 
mapping= {
    "female":0,
    "male":1,
}

print(len(df2))
df2["Sex"]= df2["Sex"].map(mapping)
print(len(df2))
df2.head()
```

```python
df = pd.DataFrame({
    "Color": ["Red", "Blue", "Green", "Red"]
})

encoded = pd.get_dummies(df["Color"])

print(encoded)
```

```python
df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40]
})

minimum = df["Age"].min()
maximum = df["Age"].max()

df["Age_normalized"] = (
    df["Age"] - minimum
) / (maximum - minimum)

print(df)
```
