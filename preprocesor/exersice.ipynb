```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("../newCourse/mizban_orders.csv",index_col=0)

value_count= df[["orders"]].value_counts()
df["value_count"]= value_count

df.head(2)
```

```python
X = df.drop("orders", axis=1)
y = df[["orders"]]
y
```

```python
def split_data(X, y , test_size=0.2, randome_state=42):
    
    random_indeces = np.random.permutation(len(X))
    
    test_set_indeces=  int(len(random_indeces) * test_size)
    
    train_size_indeces= random_indeces[test_set_indeces:]
    test_size_indeces=  random_indeces[:test_set_indeces]

    X_train, X_test= X.iloc[train_size_indeces], X.iloc[test_size_indeces]
    y_train, y_test= y.iloc[train_size_indeces], y.iloc[test_size_indeces]
    
    return X_train, X_test, y_train, y_test
split_data(X,y)
```
