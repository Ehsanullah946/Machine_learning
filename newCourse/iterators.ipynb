```python
class C_iterator:
    def __init__(self, start=0, limit=None, step = 1):
        self.start = start
        self.limit = limit
        self.step = step
    def __iter__(self):
        self.value = self.start
        return self
    def __next__(self):
        x = self.value
        if self.limit is not None:
            if self.value < self.limit:
                self.value += self.step
            else:
                raise StopIteration
        else:
            self.value +=1
        return x
```

```python
my_iterator= C_iterator(0,10,2).__iter__()   

for i in my_iterator:
    print(i)            
```

```python
class C_iterator:
    def __init__(self,iterable_object):
        self.obj = iterable_object
        self.__iter__()
    def __iter__(self):
        self.c_index = 0
        return self
    def __next__(self):
        try:
            x = self.obj[self.c_index]
            self.c_index+=1
        except:
            self.c_index-=1
            raise StopIteration
        return x
```

```python
my_iterator= C_iterator(my_list).__iter__()   

for i in my_iterator:
    print(i)   
```
