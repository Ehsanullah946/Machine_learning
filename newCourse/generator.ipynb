```python
def gen(limit):
    for i in range(limit):
        yield i 
```

```python
my_gen= gen(5)
next(my_gen)
for i in my_gen:
    print(i)
```

```python
def gen2():
    i = 0
    while True:
        yield i
        i += 1
my_gen2= gen2()
               
```

```python
next(my_gen2)
```

```python
def reading_file(file_path):
    with open(file_path) as file:
        for line in file:
            yield line.strip()

for line in reading_file("mizban_orders.csv"):
        print(line)            
    
```
