```python
import pandas as pd
import numpy as np
from matplotlib import cm
import matplotlib.pyplot as plt
```

```python
df= pd.read_table("fruit_data_with_colors.txt")
df.head()
```

```python
lookup_fruit_name= dict(zip(df.fruit_label.unique(), df.fruit_name.unique()))
print(lookup_fruit_name)     
```

```python
X = df[["mass", "width","height"]]
y= df[["fruit_label"]]

print(len(X))
def train_test_split(X,y, test_size=0.2, randome_state=42):
    random_indeces= np.random.permutation(len(X))
    
    test_set_indeces= int(len(random_indeces) * test_size) 
    
    train_size_indeces= random_indeces[test_set_indeces:]
    test_size_indeces= random_indeces[:test_set_indeces]
    
    X_train, X_test= X.iloc[train_size_indeces], X.iloc[test_size_indeces]
    y_train, y_test= y.iloc[train_size_indeces], y.iloc[test_size_indeces]
    
    return X_train,X_test, y_train, y_test
X_train,X_test, y_train, y_test=train_test_split(X,y, 0.2)   
```

```python
cmap= cm.get_cmap("gnuplot")
scatter= plt.scatter_matrix(X_train, c= y_train,marker="o", s=40, hist_kwds={"bins":15},figsize=(12,12),cmap=cmap)
```
