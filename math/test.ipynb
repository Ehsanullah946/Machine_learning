```python
def initial_parameters(x,y):
    W = np.random.randn(x,y) * 0.01
    B = np.zeros((y,1))
    
    parameters= {"W":W,"B":B}
    
    return parameters

parameters = initial_parameters(1,1)
print(parameters["W"])
print(parameters["B"])
    
     
```
