```python
def calculate_birthday_prob(n):
    if n > 365:
        return 1.0
    
    prob_different = 1.0
    for i in range(n):
        prob_different *= (365 - i) / 365
        
    return 1 - prob_different

n=360
result = calculate_birthday_prob(n)
print(f"the probablity of {n} people is {result:.3%}")
```
